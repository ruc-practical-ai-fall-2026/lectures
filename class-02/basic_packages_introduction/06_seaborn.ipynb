# `seaborn`

Within the Python ecosystem, [`seaborn`](https://seaborn.pydata.org/) provides a high level interface for statistical graphics and is used extensively in machine learning and data science work.

An introduction to `seaborn` is provided [here](https://seaborn.pydata.org/tutorial/introduction.html) along with a few examples below.

A [gallery](https://seaborn.pydata.org/examples/index.html) of additional examples and a [tutorial](https://seaborn.pydata.org/tutorial.html) are available on the official website.

## Setup

We start by importing seaborn. Note that we also import the corresponding `matplotlib` functions that we need.

Seaborn is built on top of `matplotlib` and provides convenient functions for creating statistical visualizations, while `matplotlib` provides the underlying figure, axes, and display system.

We commonly import `matplotlib.pyplot` with `seaborn` so we can control details such as titles, axis labels, figure size, and when the plot is displayed with `plt.show()`.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

We will use an basic dataset for the following examples.

```python
penguins = sns.load_dataset("penguins")
```

`seaborn` allows us to set themes and other global parameters.

```python
sns.set_theme(style="ticks", palette="pastel")
```

## Scatter Plots

We can use `seaborn` to generate scatter plots and visualize the relationships between variables. Note that we call `plt.show()` directly to show the plot.

```python
sns.scatterplot(data=penguins, x="bill_length_mm", y="bill_depth_mm")
plt.show()
```

`seaborn` makes it easy to visualize categories in classes with colors.

```python
sns.scatterplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",
)

plt.show()
```

A more involved example from the gallery:

```python
# Load the example diamonds dataset
diamonds = sns.load_dataset("diamonds")

# Draw a scatter plot while assigning point colors and sizes to different
# variables in the dataset
f, ax = plt.subplots(figsize=(6.5, 6.5))
sns.despine(f, left=True, bottom=True)
clarity_ranking = ["I1", "SI2", "SI1", "VS2", "VS1", "VVS2", "VVS1", "IF"]
sns.scatterplot(
    x="carat",
    y="price",
    hue="clarity",
    size="depth",
    palette="ch:r=-.2,d=.3_r",
    hue_order=clarity_ranking,
    sizes=(1, 8),
    linewidth=0,
    data=diamonds,
    ax=ax,
)
```

## Histograms

`seaborn` makes inspecting feature distributions with histograms easy.

```python
sns.histplot(data=penguins, x="body_mass_g", bins=20)

plt.show()
```

## Pair Plot

Pair plots are ubiquitous in exploratory data analysis (EDA). It is common practice to make a pair plot as a first step in exploring a new dataset.

```python
sns.pairplot(
    penguins,
    vars=["bill_length_mm", "bill_depth_mm", "flipper_length_mm"],
    hue="species",
)

plt.show()
```

## Where did the pair plot come from?

Pair plots, and specifically the `seaborn` implementation, are used ubiquitously in modern data science. While they are not generally attributed to any single inventor or author, they evolved out of the growth of *exploratory data analysis* (EDA) in the 1970s and 1980s into the form we know today.

John Tukey, a major figure in modern statistics and co-developer of the Cooley–Tukey Fast Fourier Transform (FFT), is generally credited with introducing the modern field of exploratory data analysis in his classic 1977 *Exploratory Data Analysis* book. While Tukey's EDA book covers multi-frame visualizations and the techniques of EDA, it does not introduce the pair plot itself, which has since become one of the more popular tools in modern EDA.

In their 2012 paper on the "generalized pairs plot" Emerson *et al.* credit J. A. Hartigan for introducing the pair plot in his 1975 paper [*Printer graphics for clustering*](https://www.tandfonline.com/doi/epdf/10.1080/0094965750881012) and John W. Tukey and Paul A. Tukey for introducing the draftsmen's display, a matrix of scatter plots for examining pairwise relationships in multi-variate data. The plot style was further popularized in the book *Graphical Methods for Data Analysis* (1983) by J. M. Chambers and through implementation in R and its predecessor [S](https://en.wikipedia.org/wiki/S_(programming_language)) developed at Bell Laboratories under Chambers. The same general idea later became standard in R and other statistical-computing environments.

Details on `seaborn`'s modern API are available [here](https://seaborn.pydata.org/generated/seaborn.pairplot.html).
